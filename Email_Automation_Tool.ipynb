```python
# Write personalizer module
personalizer_code = '''
from jinja2 import Template
from datetime import date

def load_template(template_path):
    """Load email template from file."""
    with open(template_path, "r") as f:
        return f.read()

def personalize_email(template_str, contact, subject):
    """
    Fill template placeholders with contact data.
    Returns (subject, body) tuple.
    """
    template = Template(template_str)
    body = template.render(
        name=contact["name"],
        task=contact["task"],
        department=contact["department"],
        subject=subject,
        date=date.today().strftime("%B %d, %Y")
    )
    return subject, body
'''

with open("src/personalizer.py", "w") as f:
    f.write(personalizer_code)

print("✅ personalizer.py created!")
```

```python
# Write email sender module
sender_code = '''
import smtplib
import os
import logging
from email.message import EmailMessage

def send_email(to_email, subject, body, dry_run=True):
    """
    Send email via SMTP or simulate in dry-run mode.
    Returns True on success, False on failure.
    """
    if dry_run:
        print(f"\\n[DRY-RUN] ✉️  Would send to : {to_email}")
        print(f"  Subject  : {subject}")
        print(f"  Preview  : {body[:80].strip()}...")
        logging.info(f"DRY-RUN | TO: {to_email} | SUBJECT: {subject} | STATUS: simulated")
        return True

    try:
        sender = os.getenv("EMAIL_ADDRESS")
        password = os.getenv("EMAIL_PASSWORD")

        msg = EmailMessage()
        msg["From"] = sender
        msg["To"] = to_email
        msg["Subject"] = subject
        msg.set_content(body)

        with smtplib.SMTP_SSL("smtp.gmail.com", 465) as smtp:
            smtp.login(sender, password)
            smtp.send_message(msg)

        logging.info(f"SENT | TO: {to_email} | SUBJECT: {subject} | STATUS: success")
        print(f"[SENT] ✅ Email delivered to {to_email}")
        return True

    except Exception as e:
        logging.error(f"FAILED | TO: {to_email} | ERROR: {str(e)}")
        print(f"[FAILED] ❌ Could not send to {to_email}: {e}")
        return False
'''

with open("src/email_sender.py", "w") as f:
    f.write(sender_code)

print("✅ email_sender.py created!")
```

```python
# Write reporter module
reporter_code = '''
import pandas as pd
import os

def generate_report(results, output_path="outputs/report.csv"):
    """
    Save send results to CSV report.
    results = list of dicts with name, email, subject, status
    """
    os.makedirs("outputs", exist_ok=True)
    df = pd.DataFrame(results)
    df.to_csv(output_path, index=False)
    print(f"\\n📊 Report saved to: {output_path}")
    print(df.to_string(index=False))
    return df
'''

with open("src/reporter.py", "w") as f:
    f.write(reporter_code)

print("✅ reporter.py created!")
```

```python
import os
import sys
import logging
import pandas as pd
from datetime import date
from jinja2 import Template

# ── Add src to path ──────────────────────────────────────
sys.path.insert(0, "src")

# ── Setup logging ────────────────────────────────────────
os.makedirs("logs", exist_ok=True)
os.makedirs("outputs", exist_ok=True)

logging.basicConfig(
    filename="logs/email_log.txt",
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)

# ── Import modules ────────────────────────────────────────
from personalizer import load_template, personalize_email
from email_sender import send_email
from reporter import generate_report

# ── Config ────────────────────────────────────────────────
DRY_RUN = True   # ← Change to False only for real emails

# ── Load Data ─────────────────────────────────────────────
contacts_df  = pd.read_csv("data/contacts.csv")
reminders_df = pd.read_csv("data/reminders.csv")
template_str = load_template("templates/reminder_template.txt")

# ── Merge contacts + reminders ────────────────────────────
merged_df = reminders_df.merge(
    contacts_df,
    left_on="recipient_email",
    right_on="email"
)

# ── Header ────────────────────────────────────────────────
print("=" * 55)
print("   📧 Email Automation & Reminder System")
print("=" * 55)
print(f"   Mode : {'🔵 DRY-RUN (no real emails)' if DRY_RUN else '🟢 LIVE MODE'}")
print(f"   Total contacts loaded : {len(contacts_df)}")
print(f"   Total reminders loaded: {len(reminders_df)}")
print("=" * 55)

# ── Process & Send ────────────────────────────────────────
results = []

for _, row in merged_df.iterrows():
    subject, body = personalize_email(
        template_str,
        contact={
            "name"       : row["name"],
            "task"       : row["task"],
            "department" : row["department"]
        },
        subject=row["subject"]
    )

    success = send_email(
        to_email = row["recipient_email"],
        subject  = subject,
        body     = body,
        dry_run  = DRY_RUN
    )

    results.append({
        "name"    : row["name"],
        "email"   : row["recipient_email"],
        "subject" : subject,
        "status"  : "sent" if success else "failed"
    })

# ── Report ────────────────────────────────────────────────
report_df = generate_report(results)

# ── Summary ───────────────────────────────────────────────
sent   = sum(1 for r in results if r["status"] == "sent")
failed = len(results) - sent

print(f"\n{'='*55}")
print(f"  ✅ Total Sent    : {sent}")
print(f"  ❌ Total Failed  : {failed}")
print(f"  📁 Log file      : logs/email_log.txt")
print(f"  📊 Report file   : outputs/report.csv")
print(f"{'='*55}")
```

```python
def send_test_email(to_email, sender_email, sender_password):
    """
    Send a single test email to verify SMTP works.
    Always test with ONE email before bulk sending.
    """
    try:
        msg = EmailMessage()
        msg["From"]    = sender_email
        msg["To"]      = to_email
        msg["Subject"] = "✅ Test Email from Python Automation System"
        msg.set_content(f"""
Hello!

This is a test email sent from Python using Gmail SMTP.

Sent by  : {sender_email}
Sent to  : {to_email}
Date     : {date.today().strftime("%B %d, %Y")}

Your Email Automation System is working correctly!

Best regards,
Python Automation System
        """)

        print(f"⏳ Connecting to Gmail SMTP...")

        with smtplib.SMTP_SSL("smtp.gmail.com", 465) as smtp:
            smtp.login(sender_email, sender_password)
            smtp.send_message(msg)

        print(f"✅ Test email SENT successfully to {to_email}")
        return True

    except smtplib.SMTPAuthenticationError:
        print("❌ Authentication failed!")
        print("   → Check your Gmail address is correct")
        print("   → Make sure you used APP PASSWORD not regular password")
        print("   → Check 2-Step Verification is enabled")
        return False

    except smtplib.SMTPException as e:
        print(f"❌ SMTP Error: {e}")
        return False

    except Exception as e:
        print(f"❌ Unexpected error: {e}")
        return False


# ── Run Test ──────────────────────────────────────────────
# Send test email to yourself first
test_recipient = input("📬 Enter recipient email for test (use your own): ")

result = send_test_email(
    to_email        = test_recipient,
    sender_email    = EMAIL_ADDRESS,
    sender_password = EMAIL_PASSWORD
)

if result:
    print("\n🎉 SMTP connection working! Ready to send bulk emails.")
else:
    print("\n⚠️  Fix the error above before proceeding.")
```

```python
import time

# Setup logging
os.makedirs("/content/logs", exist_ok=True)
os.makedirs("/content/outputs", exist_ok=True)

logging.basicConfig(
    filename = "/content/logs/email_log.txt",
    level    = logging.INFO,
    format   = "%(asctime)s | %(levelname)s | %(message)s",
    force    = True
)

def send_real_email(to_email, subject, body, sender_email, sender_password):
    """Send one personalized email via Gmail SMTP."""
    try:
        msg = EmailMessage()
        msg["From"]    = sender_email
        msg["To"]      = to_email
        msg["Subject"] = subject
        msg.set_content(body)

        with smtplib.SMTP_SSL("smtp.gmail.com", 465) as smtp:
            smtp.login(sender_email, sender_password)
            smtp.send_message(msg)

        logging.info(f"SENT | TO: {to_email} | SUBJECT: {subject} | STATUS: success")
        return True

    except Exception as e:
        logging.error(f"FAILED | TO: {to_email} | ERROR: {str(e)}")
        return False


# ── Confirmation before sending ───────────────────────────
print("⚠️  YOU ARE ABOUT TO SEND REAL EMAILS")
print(f"   Total emails : {len(merged_df)}")
print(f"   Sending from : {EMAIL_ADDRESS}")
print()
confirm = input("Type YES to confirm and send: ")

if confirm.strip().upper() != "YES":
    print("❌ Cancelled. No emails sent.")
else:
    results = []

    print("\n🚀 Sending emails...\n")

    for i, (_, row) in enumerate(merged_df.iterrows(), start=1):

        subject, body = personalize_email(
            template_str,
            contact = {
                "name"       : row["name"],
                "task"       : row["task"],
                "department" : row["department"]
            },
            subject = row["subject"]
        )

        success = send_real_email(
            to_email        = row["recipient_email"],
            subject         = subject,
            body            = body,
            sender_email    = EMAIL_ADDRESS,
            sender_password = EMAIL_PASSWORD
        )

        status = "sent" if success else "failed"
        icon   = "✅" if success else "❌"

        print(f"[{i:02d}/20] {icon} {row['name']:20s} → {row['recipient_email']}")

        results.append({
            "no"         : i,
            "name"       : row["name"],
            "email"      : row["recipient_email"],
            "department" : row["department"],
            "subject"    : subject,
            "status"     : status
        })

        # ⏳ Small delay to avoid Gmail rate limit
        time.sleep(1.5)

    # ── Save Report ───────────────────────────────────────
    report_df = pd.DataFrame(results)
    report_df.to_csv("/content/outputs/report.csv", index=False)

    # ── Summary ───────────────────────────────────────────
    sent   = report_df[report_df["status"] == "sent"].shape[0]
    failed = report_df[report_df["status"] == "failed"].shape[0]

    print(f"\n{'=' * 55}")
    print(f"  ✅ Successfully Sent : {sent}")
    print(f"  ❌ Failed            : {failed}")
    print(f"  📊 Report saved      : /content/outputs/report.csv")
    print(f"  📄 Log saved         : /content/logs/email_log.txt")
    print(f"{'=' * 55}")
```
